```python
import seaborn as sns
import json
from datetime import datetime
import pandas as pd
import matplotlib.pyplot as plt


f = open("./moex_values.json", 'r')
moex_values = json.loads(f.read())
# moex_values = {k: v for k, v in moex_values.items()}

df = pd.DataFrame(list(moex_values.items()), columns=["date", "value"])

sns.lineplot(data=df, sizes=(.20, .20))
plt.show()
```
